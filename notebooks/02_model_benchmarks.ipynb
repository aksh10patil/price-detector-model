# 📊 Model Benchmarking: From Classical ML to Frontier LLMs

In this notebook, we evaluate multiple pricing models against a held-out test set:
- Constant Mean Pricer
- Linear Regression (Tabular Metadata)
- NLP + Ridge Regression (Bag-of-Words)
- Random Forest & XGBoost Ensembles
- PyTorch Deep Residual Neural Network (ResNet)
- Human Expert Benchmark
- Frontier LLMs (GPT-5 Nano, Claude, Gemini)


```python
from pricer.data.schema import Item
from pricer.models.baselines import (
    ConstantPricer,
    LinearRegressionPricer,
    NLPLearningPricer,
    RandomForestPricer,
    XGBoostPricer,
    HumanPricer,
)
from pricer.evaluation import evaluate, plot_scatter, plot_error_trend, plot_leaderboard
from pricer.config import DEFAULT_LITE_DATASET

train, val, test = Item.from_hub(DEFAULT_LITE_DATASET)
test_sample = test[:200]
```

## 1. Evaluating Classical ML Baselines


```python
# Constant Baseline
constant_model = ConstantPricer(train)
res_const = evaluate(constant_model, test_sample, title='Constant Pricer')

# NLP + Ridge
nlp_model = NLPLearningPricer(max_features=1000)
nlp_model.fit(train)
res_nlp = evaluate(nlp_model, test_sample, title='NLP + Ridge')

# Random Forest
rf_model = RandomForestPricer(n_estimators=50)
rf_model.fit(train)
res_rf = evaluate(rf_model, test_sample, title='Random Forest')

# Human Expert Benchmark
human_model = HumanPricer(test)
res_human = evaluate(human_model, test_sample, title='Human Expert (Ed)')
```

## 2. Interactive Performance Visualizations


```python
# Scatter Plot: Predicted vs Actual with Ideal y=x
fig_scatter = plot_scatter(res_nlp)
fig_scatter.show()

# Running Error Trend with 95% Confidence Interval
fig_trend = plot_error_trend(res_nlp)
fig_trend.show()
```
